```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('titanic.csv')
df
```

```python
df.info()
# sex(성별)이 object(글자) --> 숫자로 변경
# age(나이)의 결측치 263개를 채우자
# fare(운임)의 결측치 1개는 버리자
```

```python
# sex을 숫자로
df['sex2'] = df['sex'].map( {'female':1, 'male':0} ) # 1줄
df
```

```python
df[ df['fare'].isna() ] # 운임이 결측치인 1명, "Storey, Mr. Thomas"
```

```python
# 운임의 결측치 삭제
df = df.dropna(subset='fare') # drop : 버리다, na : 결측치, 1줄
df # 1309 - 1 ==> 1308
```

```python
# age 의 결측치를 채우자
```

```python
# 2-1. name 컬럼에서 호칭을 뽑는다. 예) Miss. --> Miss
df['title'] = df['name'].str.extract(r"([A-Za-z]+)\.")
df
```

```python
# df['title'].value_counts()
```

```python
# df.groupby('title')['age'].median()
```

```python
# 2-2. 호칭별 나이의 중앙값을 얻는다.
val = df.groupby('title')['age'].transform('median')
val
```

```python
# 2-3. 결측치를 채운다.
df['age'] = df['age'].fillna(val) # fill : 채운다, na : 결측치
```

```python
df['age'].isna().sum() # age    0 <-- 성공
```

```python
# 요약 5줄
# 1.   df['sex2'] = df['sex'].map( {'female':1, 'male':0} )   # 성별을 숫자로
# 3.   df = df.dropna(subset='fare')                          # 운임 결측치 삭제
# 2-1. df['title'] = df['name'].str.extract(r"([A-Za-z]+)\.") # 호칭 뽑기
# 2-2. val = df.groupby('title')['age'].transform('median')   # 중앙값 얻기
# 2-3. df['age'] = df['age'].fillna(val)                      # 나이의 결측치 채우기
```

```python
# 저장
df.to_csv('titanic_ok.csv', index=False) # 인덱스는 빼고 저장
```
