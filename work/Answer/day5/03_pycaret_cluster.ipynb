```python
import pandas as pd
```

```python
df = pd.read_csv('../../data/pycaret_data/jewellery.csv')
df
```

```python
from pycaret.clustering import *
```

```python
setup(df, normalize=True)
```

```python
kmeans = create_model('kmeans')
```

```python
kmeans
```

```python
evaluate_model(kmeans)
```

```python
plot_model(kmeans, plot='elbow')
```

```python
plot_model(kmeans, plot='silhouette')
```
