## 4. 다중공선성: VIF 📋

VIF가 10 이상이면 다른 입력들과 정보가 많이 겹친다는 신호입니다.

```python
X_const = sm.add_constant(X_train)
vif = {}
for i, col in enumerate(X_train.columns, start=1):     # 0번은 상수항이라 1번부터
    vif[col] = variance_inflation_factor(X_const.values, i)
pd.Series(vif).sort_values(ascending=False).round(1)
```
